```python
import pandas as pd 
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer 


from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error 
```

```python
data_1 = pd.read_csv("student-por.csv")
data_2 = pd.read_csv("student-mat.csv")
```

```python
data_1.corr(numeric_only= True)["G3"].sort_values(ascending = False)
```

```python
data_2.corr(numeric_only= True)["G3"].sort_values(ascending = False)
```

```python
data_1.shape
```

```python
train_labels = data_1["G3"]
train_features = data_1.drop("G3", axis = 1)
```

```python
test_labels = data_2["G3"]
test_features = data_2.drop("G3", axis = 1)
```

```python
train_cat = train_features.drop(["age", "Medu",  "Fedu", "traveltime", "studytime", "failures", "famrel", "freetime", "goout", "Dalc", "Walc", "health", "absences", "G1", "G2"], axis = 1).columns.tolist()

train_num = train_features.drop(["school", "sex", "address","famsize","Pstatus", "Mjob", "Fjob", "reason", "guardian", "schoolsup", "famsup", "paid", "activities", "nursery", "higher", "internet", "romantic" ], axis = 1).columns.tolist()

```

```python
test_cat = test_features.drop(["age", "Medu",  "Fedu", "traveltime", "studytime", "failures", "famrel", "freetime", "goout", "Dalc", "Walc", "health", "absences", "G1", "G2"], axis = 1).columns.tolist()

test_num = test_features.drop(["school", "sex", "address","famsize","Pstatus", "Mjob", "Fjob", "reason", "guardian", "schoolsup", "famsup", "paid", "activities", "nursery", "higher", "internet", "romantic" ], axis = 1).columns.tolist()


```

```python
num_pipeline = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
    ])


cat_pipeline = Pipeline([
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

full_pipeline = ColumnTransformer([
    ("num", num_pipeline, train_num), 
    ("cat", cat_pipeline, train_cat)
])

```

```python
student_train_features  = full_pipeline.fit_transform(train_features)
# train_lebels
                                                     
student_test_features  = full_pipeline.fit_transform(test_features)
# test_labels
```

```python
# to predict linear regressor
lin_reg = LinearRegression()
lin_reg.fit(student_train_features, train_labels)
# to precit to data
lin_regs = lin_reg.predict(student_test_features)
#print(lin_regs)

# rmse 
lin_rmse = root_mean_squared_error( test_labels, lin_regs)
print(lin_rmse)
```

```python
# to predict Randome forest regresor
ran_reg = RandomForestRegressor()
ran_reg.fit(student_train_features, train_labels)
# to predict the data

ran_regs = ran_reg.predict(student_test_features)
#print(ran_regs)

# to check root mean
ran_rmse = root_mean_squared_error(test_labels, ran_regs)
print(ran_rmse)
```

```python
# to predict decisionteeregressor

dec_reg =DecisionTreeRegressor()
dec_reg.fit(student_train_features, train_labels)
dec_regs = dec_reg.predict(student_test_features)
#print(dec_regs)

# rmse 
dec_rmse = root_mean_squared_error(test_labels, dec_regs)
print(dec_rmse)
```

```python
# to predict GradientBoostingRegressor
gra_reg = GradientBoostingRegressor()
gra_reg.fit(student_train_features, train_labels)
gra_regs = gra_reg.predict(student_test_features)
#print(gra_regs)

# rmse 
gra_rmse = root_mean_squared_error(test_labels, gra_regs)
print(gra_rmse)
```
